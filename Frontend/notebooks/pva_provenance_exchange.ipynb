## Components — Responsibilities
- **ERC-721 Deed**: Master ownership token; anchors physical custody and transfers; enforces royalties (EIP-2981).
- **ERC-20 Shares**: Fractional financial stake; priced by bonding curve; tradable on DEX (Uniswap V3).
- **Bonding Curve**: Automatic price function; increases with net buys, decreases with net sells; 24/7 live auction.
- **PVACoin (ERC-20)**: Utility token for fees and marketplace actions; liquid pairs ($PVA/ETH, $PVA/USDC).
- **Provenance DB/Indexing**: On-chain events indexed (The Graph) to render human-readable timelines.
- **Vault & Custody**: Physical storage/custodianship; QR ties asset to its live page; photo proof-of-custody (IPFS).
- **Marketplace DApp**: Mint/list/buy/sell; handles fractionalization, price quotes, wallet integration (MetaMask/WalletConnect).
- **Fiat Bridge**: Off-chain service (Circle/Stripe) converts stablecoins to fiat and triggers bank payouts (oracle flow).
- **Social Lounges**: Token-gated rooms per asset (Guild.xyz/Collab.Land); tiers for observers, contributors, creators.

## Smart Contracts — Modules
- **Deed Contract (ERC-721)**: mint, transfer, tokenURI; setRoyaltyInfo; emit `DeedTransferred`.
- **Fractional Shares (ERC-20)**: mint/burn per asset pool; bonding curve buy/sell; emit `SharesTraded`.
- **Bonding Curve**: `price(n)` function; spread/fee; slippage controls; circuit breakers.
- **Referral Rewards**: one-time per item per referrer; on purchase path with referrer param; rate-limited.
- **Treasury & Fees**: split fees (artist/PVA/treasury); staking rewards for $PVA.
- **DAO Governance**: proposals, votes, parameter updates; multi-sig for upgrades; timelocks.

## Off-Chain Oracle — Responsibilities
- Listen to events: `ItemPurchased`, `CustodyProofSubmitted`, `FiatPayoutRequested`.
- Trigger emails/SMS (Twilio/SendGrid) for order and shipment steps.
- Initiate fiat conversions/payouts via Circle API; confirm on-chain state with `confirmFiatPayment()`.
- Store custody photos and metadata on IPFS; write CID to Deed history.
- Audit logging, retries, and secure secrets management (KMS/Env).

## QR Link & Live Page
- **QR**: Encodes canonical URL (e.g., `/asset/PVA-RUBY-001`) + optional `referrerWallet` param.
- **Live page**: Story, media, current status (vaulted/available), fractional market (price, supply, holders), custody timeline.
